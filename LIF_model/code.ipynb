```python
# import libraries
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
```

```python
# input parameters
v_rest = -65.0
v_threshold = -50.0
v_reset = -70.0
tau = 10.0
R= 10.0
dt = 0.1
T = 200.0
```

# DEFINE FUNCTIONS =====================================

```python
def simulations(I):
    steps = int(T/dt)
    V = np.zeros(steps)
    V[0] = v_rest
    spikes = []
    for t in range(1,steps):
        dV = (-(V[t-1] - v_rest) + R * I) /tau
        V[t] = V[t-1] + dV * dt
        if V[t] >= v_threshold:
            spikes.append(t*dt)
            V[t] = v_reset
    return V, spikes
```

```python
# Single run with constant input current
V, spikes = simulations(2.0)
print("Number of spikes at 2.0 nA :", len(spikes))
time = np.arange(len(V)) *dt
plt.plot(time,V)
plt.xlabel('Time (ms)')
plt.ylabel('Membrane Voltage (mV)')
plt.savefig('membrane_voltage.png')
plt.close()
```

```python
# F-i curve
currents = np.linspace(0,5,50)
rates = [len(simulations(I)[1])/(T/1000) for I in currents]
for I, r in zip(currents[::5], rates[::5]):
    print(f"Current: {I:.2f} nA, Firing Rate: {r:.2f} Hz")
plt.plot(currents, rates)
plt.xlabel('Input Current (nA)')
plt.ylabel('Firing Rate (Hz)')
plt.savefig('f_i_curve.png')
```
